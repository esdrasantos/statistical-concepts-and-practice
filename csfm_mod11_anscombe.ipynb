```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sn
import scipy.stats as stats
```

```python
from google.colab import drive

drive.mount('/content/drive')
```

```python
anscombe = pd.read_excel('/content/drive/MyDrive/Curso de Estatística 2.0/Arquivos Excel/Dados4_CursoEstatistica.xlsx', sheet_name='Anscombe')

anscombe.head()
```

```python
# REGRESSÃO CONJUNTO X1Y1

X = sn.add_constant(anscombe['X1'])
Y = anscombe['Y1']

model = sn.OLS(Y, X)
results = model.fit()
print("Regressão X1Y1", results.summary())
```

```python
# REGRESSÃO CONJUNTO X2Y2

X = sn.add_constant(anscombe['X2'])
Y = anscombe['Y2']

model = sn.OLS(Y, X)
results = model.fit()
print("Regressão X2Y2", results.summary())
```

```python
# REGRESSÃO CONJUNTO X3Y3

X = sn.add_constant(anscombe['X3'])
Y = anscombe['Y3']

model = sn.OLS(Y, X)
results = model.fit()
print("Regressão X3Y3", results.summary())
```

```python
# REGRESSÃO CONJUNTO X4Y4

X = sn.add_constant(anscombe['X4'])
Y = anscombe['Y4']

model = sn.OLS(Y, X)
results = model.fit()
print("Regressão X4Y4", results.summary())
```

Ou seja, todos as mesmas métricas...
Entretanto, não são iguais e isso podemos verificar através dos gráficos de dispersão

```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(7, 7))

axes[0, 0].scatter(anscombe['X1'], anscombe['Y1'])
axes[0, 0].set_title('Conjunto X1Y1')
axes[0,1].scatter(anscombe['X2'], anscombe['Y2'])
axes[0, 1].set_title('Conjunto X2Y2')
axes[1,0].scatter(anscombe['X3'], anscombe['Y3'])
axes[1, 0].set_title('Conjunto X3Y3')
axes[1,1].scatter(anscombe['X4'], anscombe['Y4'])
axes[1, 1].set_title('Conjunto X4Y4')

plt.show()
```
