# Kruskal-Wallis (3+ Médias - Medianas)

```python
import pandas as pd
import numpy as np
```

```python
notas = pd.read_excel('/content/drive/MyDrive/Curso de Estatística 2.0/Arquivos Excel/Dados3_CursoEstatistica.xlsx', sheet_name=6)

notas.head()
```

```python
notas_g = notas.groupby('Etnia')['Nota'].apply(list)

notas_g.head()
```

```python
from scipy.stats import kruskal

stat, p = kruskal(*notas_g)

print('stat=%.4f, p=%.4f' % (stat, p))
```

```python
# Decisão:

alpha = 0.05

if p > alpha:
  print('Não rejeitamos a hipótese nula')
  print('Logo, as medianas são iguais')
else:
  print('Rejeitamos a hipótese nula')
  print('Logo, existe ao menos uma mediana diferente')
```
