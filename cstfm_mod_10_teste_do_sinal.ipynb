```python
import pandas as pd
import numpy as np
```

```python
dominos = pd.read_excel('/content/drive/MyDrive/Curso de Estatística 2.0/Arquivos Excel/Dados3_CursoEstatistica.xlsx', sheet_name=7)

dominos.head()
```

```python
import statsmodels.api as sm

dominos['Diff'] = dominos['Antiga'] - dominos['Nova']

dominos.head()
```

```python
# Contar os sinais +

positivos = dominos[dominos['Diff'] > 0].sum()
positivos
```

```python
# Numero de observações validas (não nulas)

n = len(dominos[dominos['Diff'] != 0])
n
```

```python
# Teste binomial
# Utilizado quando se tem duas opções de valor: sucesso e insucesso (+ ou -)

p = 0.5 # Probabilidade de sucesso (+)

pvalor = sm.stats.binom_test(len(positivos), n, p, alternative='smaller')

pvalor
```

```python
# Decisão
alfa = 0.05

if pvalor < alfa:
  print('Rejeitar H0')
  print('Portanto, as proporções dos sinais são estatisticamente diferentes')
  print('Nesse caso, em específico, o a proporção de Antiga e menor que a proporção de Nova')

else:
  print('Aceitar H0')
  print('Portanto, as proporções dos sinais são estatisticamente iguais')
  print('Nesse caso, em específico, o a proporção de Antiga e maior ou igual que a proporção de Nova')
```
